```python
#Solicite ao usuário um número inteiro positivo.utilizando um laço de repetição, calcule e exiba a soma de
#todos os números de 1 até o número informado.

numero = int(input("Digite um número: "))
soma = 0

for i in range(1, numero + 1):
 soma += i

 print(f"Resultado:{soma}")
```

```python
#Um caixa eletrônico possui notas de:
#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1
#Solicite ao usuário o valor do saque e informe a menorquantidade possível de notas para realizar o pagamento.

valor = int(input("Digite o valor do saque: "))
nota = 100

while nota >= 1:
    quantidade = valor // nota

    if quantidade > 0:
        print(quantidade, "nota(s) de R$", nota)

    valor = valor % nota

    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    elif nota == 2:
        nota = 1
    else:
        break
```
