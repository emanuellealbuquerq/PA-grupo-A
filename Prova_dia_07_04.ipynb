Classificação de idade

```python
#entrada

nome = str(input("digite seu nome: "))
idade = int(input("digite sua idade: "))

#Processamento/Saida

if idade <= 17:
  print(f"{nome},voce tem {idade} anos de idade,voce é menor de idade")

elif idade <= 59 :
  print(f"{nome},voce tem {idade} anos de idade,voce é adulto")

else:
  print(f"{nome},voce tem {idade} anos de idade,voce é idoso")
```

Cálculo de desconto

```python
#Entrada

valor = int(input("digite um numero: "))

#Processamento/saida

if valor <= 100 :

  print(f"se o valor for de {valor},ele não tem desconto")

elif valor <= 500 :

  print(f"se o valor for de {valor},ele tem 10% de desconto")

else:
  print(f"se o valor for de {valor},ele tem 20% de desconto")
```

Verificação de triângulo

```python
#Entrada

lado1 = int(input("digite o lado do triangulo: "))
lado2 = int(input("digite o lado do triangulo: "))
lado3 = int(input("digite o lado do triangulo: "))

#Verifique se o triangilo é verdadeiro

if lado1 + lado2 > lado3 and lado1 + lado3 > lado2 and lado2 + lado3 > lado1
  print(f"tringulo válido")
```
