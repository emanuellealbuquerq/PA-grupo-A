```python
#Jogos inteligente

# 1 -par/impar
# 2 -positivo
# 3 -maior de 02 numeros
# 4 -sair

print("___________Jogo Inteligente___________")
print("opção 1: par/impar")
print("opção 2: positivo")
print("opção 3: maior de 02 números")
print("opção 4: sair")
print("_______________________________________")

opcao = int(input("Digite a opção desejada: "))

if opcao == 1:
    num = int(input("Digite um número: "))
    if num % 2 == 0:
        print("O número é par!")
    else:
        print("O número é ímpar!")

elif opcao == 2:
    num = int(input("Digite um número: "))
    if num > 0:
        print("O número é positivo!")
    elif num < 0:
        print("O número é negativo!")
    else:
        print("O número é zero!")

elif opcao == 3:
    num1 = int(input("Digite o primeiro número: "))
    num2 = int(input("Digite o segundo número: "))
    if num1 > num2:
        print("O primeiro número é maior!")
    elif num1 < num2:
        print("O segundo número é maior!")
    else:
        print("Os números são iguais!")

elif opcao == 4:
    print("Saindo do jogo...")
```

```python
#Escrever um programa que verifique a senha,o usuario tera 03 tentativas
#caso erre as 3 imprimir mensagem de "acesso negado e tente novamente mais tarde" e encerrar o programa.

senha = ""
tentativas = 3


while senha != "12190" and tentativas > 0:
    senha = input("Digite a senha: ")
    tentativas -= 1

    if senha == "12190":
        print("Acesso permitido!")

    elif tentativas == 0:
        print("Acesso negado! Tente novamente mais tarde!")
    else:
        print(f"Acesso negado! Você ainda tem {tentativas} tentativas.")
```
