```python
#Escrever um programa que o usuário digita um número de 1 a 10 e o programa imprimi a tabuada do número digitado.

numero = int(input("Digite um número de 1 a 10: "))

for i in range (0,11):
  print(numero * i)
```

```python
#Escrever um programa que o usuário digita um número e o progranma imprimi até o número digitado

numero = int(input("Digite um número: "))

for i in range (numero + 1):
  print(i)
```

```python
#Escrever um programa que o usuário digita um número e o programa imprimr o número digitado até o zero (contegem regressiva)

numero = int(input("Digite um número: "))

for i in range (numero):
  print(numero - i)
```

```python
#Escrever um programa que o usuário digita um número e o programa imprimi os números impares

numero = int(input("Digite um número: "))

for i in range (0,numero - 1):
  print(i)
```
