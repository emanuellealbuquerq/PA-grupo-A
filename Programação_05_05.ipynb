Exercicio 0.1 - Escrever um programa que o usuário digita um número de 1 a 10 e o programa imprimi a tabuada do número digitado.


```python
#Escrever um programa que o usuário digita um número de 1 a 10 e o programa imprimi a tabuada do número digitado.

numero = int(input("Digite um número de 1 a 10: "))

for i in range (0,11):
  print(numero * i)
```

Exercicios 2 - Escrever um programa que o usuário digita um número e o programa imprimr o número digitado até o zero (contegem regressiva)


```python
#Escrever um programa que o usuário digita um número e o programa imprimr o número digitado até o zero (contegem regressiva)

numero = int(input("Digite um número: "))

for i in range (numero):
  print(numero - i)
```

Exercicio 9 - Escrever um programa que o usuário digita 04 números e o programa imprimi o maior


```python
#Escrever um programa que o usuário digita 04 números e o programa imprimi o maior

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))
num4 = float(input("Digite o quarto número: "))

numeros = [num1, num2, num3, num4]
maior = max(numeros)

print(f"O maior número digitado é: {maior}")
```

Exercicio 10 - Escrever um programa que receba um numero N e imprima uma piramide perfeitamente centralizada

```python
#Escrever um programa que receba um numero N e imprima uma piramide perfeitamente centralizada

palavra = input("Digite um número: ")
numero = int(palavra)

for i in range(1, numero + 1):
    print(" " * (numero - i) + str(i) * (2 * i - 1))
```
