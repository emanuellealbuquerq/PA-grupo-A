```python
#Escreva um programa que o usuario digita uma palavra e verifique se a palavra é palindroma

palavra = str(input("digite uma palavra: "))
invertida = " "

for i in (palavra):

  invertida = i + invertida

if invertida == palavra:
  print(f"{palavra} é uma palavra palindroma")

else:
  print(f"{palavra} não é palavra palindroma")
```
