Par e impar

```python
#Entrada

numero = int(input("digite um numero: "))

#processamento/saida

if numero % 2 == 0:
  print(f"O numero {numero} é par.")

else:
  print(f"O numero {numero} é impar.")
```

Positivo,negativo ou zero

```python
#Entrada

numero = float(input("digite um numero: "))

#Processamento/saida

if numero == 0:
  print(f"seu  numero {numero} é zero.")

elif numero < 0:
  print(f"seu numero {numero} é negativo.")

else:
  print(f"seu numero {numero} é positivo.")

```

Calculadora simples

```python
#Entrada
a = int(input("digite um numero: "))
b = int(input("digite um numero: "))
operacao = str(input("digite uma operacao: (+,-,*,/) "))

#Processamento/saida

if operacao == '+':
  print(f"o resultado da soma de {a} e {b} é {a + b}.")

elif operacao == '-':
  print(f"o resultado da subtracao de {a} e {b} é {a - b}.")

elif operacao == '*':
   print(f"o resultado da multiplicacao de {a} e {b} é {a * b}.")

elif operacao == '/':
   if b != 0:
     print(f"o resultado da divisao de {a} e {b} é {a / b}.")
   else:
     print("Erro: Divisão por zero não é permitida.")

else:
  print("Operação inválida. Por favor, use +, -, * ou /.")
```
