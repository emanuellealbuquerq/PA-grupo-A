```python
#Escrever um programa que recebe as notas dos 04 bimestre e imprime as media com o resultado de aprovado,recupera ção ou reprovado
#Regras:Utilizar while
#Nota igual ou maior que 7 aprovado;igual ou maior que 5 recuperação;menor que 5 recuperação.)

nota = 0
bimestre = 1
resultado = 0

while bimestre <= 4:
  nota = float (input("Digite sua nota do {bimestre}º bimestre: "))
  bimestre +=1
  resultado = resultado + nota
print(resultado)

if (resultado /4) >= 7:
  print(f"Sua média foi {resultado /4},voce está aprovado.")
elif (resultado /4) >=5:
  print(f"Sua média foi {resultado /4},voce está em recuperação.")
else:
  print(f"Sua média foi {resultado /4},voce está reprovado.")
```
