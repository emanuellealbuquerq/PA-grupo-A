```python
#Escrever um programa que recebe as notas dos 04 bimestre e imprime as media com o resultado de aprovado,recupera ção ou reprovado
#Regras:Utilizar while
#Nota igual ou maior que 7 aprovado;igual ou maior que 5 recuperação;menor que 5 recuperação.)

nota = 0
bimestre = 1
resultado = 0

while bimestre <= 4:
  nota = float (input("Digite sua nota do {bimestre}º bimestre: "))
  bimestre +=1
  resultado = resultado + nota
print(resultado)

if (resultado /4) >= 7:
  print(f"Sua média foi {resultado /4},voce está aprovado.")
elif (resultado /4) >=5:
  print(f"Sua média foi {resultado /4},voce está em recuperação.")
else:
  print(f"Sua média foi {resultado /4},voce está reprovado.")
```

```python
#Projeto caixa eletronico
#Voce foi contratado para desenvolver um sistema simples de caixa eletronico
#O sistema devéra funcionar através de um menu interativo utilizando "while"
#O programa deve iniciar com um saldo inicial de R$ 10000,00 e permitir que o usuário escolha diferentes operações até decidir sair do sistema

saldo = 1000.00
opcao = 0

while opcao != 4:
    print("Caixa eletrônico")
    print("opção 1: Consultar saldo")
    print("opção 2: Depositar")
    print("opção 3: Sacar")
    print("opção 4: Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        print(f"Seu saldo atual é: R${saldo}")

    elif opcao == 2:
        deposito = float(input("Digite o valor do depósito: R$"))

        if deposito > 0:
            saldo += deposito
            print("Depósito realizado com sucesso!")
        else:
            print("Valor inválido.")

    elif opcao == 3:
        saque = float(input("Digite o valor do saque: R$"))

        if saque <= saldo:
            saldo -= saque
            print("Saque realizado com sucesso!")
        else:
            print("Saldo insuficiente.")

    elif opcao == 4:
        print("saindo do sistema")

    else:
        print("Opção inválida. Tente mais uma vez!")

print("Estarei a sua disposição!")
```
